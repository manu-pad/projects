Foi escolhido especificamente a secção de "Últimas Notícias", por ter um grande fluxo de notícias e uma boa estruturação HTML/CSS para o processo de recolha e sobre a distinção dos dados e artigos. 
https://www.sapo.pt/noticias/ultimas

O grupo precedeu então, ao desenvolvimento do código em linguagem Python para a recolha destes dados, usando como base código lecionado em aulas e produzindo um código final, testado inicialmente em Google Colabboratory, mas devido a uma maior complexidade sobre os Logs, foi a partir daí, em sua amioria desenvolvido e testado em Jupyter Notebook.

```python
import time
import json
import logging
from random import randint
import requests
from bs4 import BeautifulSoup

logging.basicConfig(
                    level=logging.INFO,
                    filename="logs.log",
                    filemode="w", 
                    format="%(asctime)s; Line:%(lineno)s; %(levelname)s: %(message)s",
                    datefmt="%d-%b-%Y %H:%M")

url = "https://www.sapo.pt/noticias/ultimas?pagina=1"
page = requests.get(url, headers={"User-Agent": "Defined"})
soup = BeautifulSoup(page.content, 'html.parser')

last_page = soup.find('li', class_='last-page')
if last_page:
    n_pages = int(last_page.get_text())
else:
    n_pages = 1

headers = {"User-Agent": "Mozilla/5.0"}
news_list = []
start_time = time.time()
run_time = 120 * 60 * 60  
news_id = 1

while time.time() - start_time < run_time:
    totalNews = 0

    for page_num in range(n_pages, 0, -1):
        url = f"https://www.sapo.pt/noticias/ultimas?pagina={page_num}"
        response = requests.get(url, headers=headers)
        soup = BeautifulSoup(response.content, 'html.parser')

        articles = soup.find_all('article', class_='article')

        if not articles:
            logging.error("Nenhum artigo encontrado. Parando")
            break

        for article in articles:
            title_tag = article.find('h3', class_='title')
            title = title_tag.text.strip() if title_tag else "Sem título"

            date_tag = article.find('span', class_='date')
            date = date_tag.text.strip() if date_tag else "Sem data"

            summary_tag = article.find('div', class_='excerpt')
            summary = summary_tag.text.strip() if summary_tag else "Sem resumo"

            link_tag = article.find('a')
            link = link_tag['href'] if link_tag else "#"
            if not link.startswith("http"):
                link = "https://www.sapo.pt" + link

            img_tag = article.find('picture')
            img = img_tag['data-original-src'] if img_tag else None

            news_item = {
                'id': news_id,
                'page': page_num,
                'title': title,
                'date': date,
                'summary': summary,
                'url': link,
                'img': img
            }

            if not any(news['url'] == news_item['url'] for news in news_list):
                news_list.append(news_item)
                totalNews += 1
                news_id += 1

        time.sleep(randint(1, 3))

    with open("noticias.json", "w", encoding="utf-8") as f:
        json.dump(news_list, f, ensure_ascii=False, indent=4)

    if totalNews == 0:
      logging.warning("Nenhuma nova notícia encontrada.")
    
    logging.info(f"Atualização concluída! {totalNews} novas notícias adicionadas.")
    time.sleep(3600)
```

Uma das dificuldades encontradas, foi garantir a iteração das páginas de notícias na SAPO; Apesar do código buscar a última notícia por conceitos de eficácia, as páginas de notícias da SAPO seguem apenas até a décima página.

A biblioteca BeautifulSoup é usada para extrair e manipular dados de ficheiros de forma simples e eficiente.

Em cada percurso de extração, o código percorre as páginas e identifica os elementos 'article' extraindo a seguir os dados:
 -título
 -data (data e hora de publicação)
 -resumo (pequena explicação sobre a notícia)
 -url (o link da notícia)
 -imagem ( o link para a imagem)

Os dados são armazenados, primeiramente em um dicionário Python que depois é guardado em um ficheiro JSON, chamado de noticias.json. Para garantir a inexistência de repetições, foi usada uma verificação na URL.

O processo é feito de novo a cada uma hora. O processo, como também definido no código, executou por 120 horas ou 5 dias.

O código também define configurações relacionadas ao ficheiro de Logging. Este ficheiro será responsável por anotar todas as chamadas e a hora em que foram realizadas, bem como garantir se algum erro fosse encontrado. Tudo ficou armazendado em um ficheiro chamado logs.log.



O código correu em uma máquina virtual pessoal durante cinco dias ininterruptos, resultando na recolha de um total de 8.429 notícias.

Indexação dos Dados no Redis

```python
from redis import Redis
from redis.commands.search.field import TextField
from redis.commands.search.indexDefinition import IndexDefinition, IndexType

# conexão do redis com o docker através do localhost
r = Redis(host='localhost', port=6379)
```

```python
# cria uma função para criar um index
def create_index(r, index_name, doc_prefix, fields):    
    try:
        r.ft(index_name).info()
        print("Index already exists!")
    except:
        r.ft(index_name).create_index(fields=fields, definition=IndexDefinition(prefix=[doc_prefix]))
        
    print("Index created")
```

A função `create index` cria o índice usando a variável do nome `index_name`, `doc_prefix` para definir um prefixo para as chaves dos documentos e a variável `fields` para os campos que serão indexados. 

```python
#função pra receber o ficheiro
def index_documents(r, doc_prefix, documents):
    for i, doc in enumerate(documents):
        doc_cleaned = {key: (value if value is not None else "") for key, value in doc.items()}
        key = f"{doc_prefix}{str(i)}"
        r.hset(key, mapping=doc_cleaned)
    
    print("Documents indexed")
```

A função `index_documents` recebe o ficheiro JSON através de `documents`. A fuhnção percorre cada objeto dentro do ficheiro e substitui valores None por strings vazias, adiciona uma chave única para cada um. A seguir, a função `hset`, armazena esses documentos no Redis.

```python
#importação do ficheiro noticias.json
import json

with open("noticias.json", "r", encoding="utf-8") as f:
    data = json.load(f)
```

Processo para a importação do ficheiro noticias.json que será utilizado ao longo do projeto como `data`.

```python
index_name = 'idx:saponoticias'
doc_prefix = 'noticias:'

fields = (
    TextField("title"),
    TextField("summary"),
    TextField("date"),
    TextField("url"),
    TextField("img")
)

create_index(r, index_name, doc_prefix, fields)
index_documents(r, doc_prefix, data)
```

<h3> Pesquisa por Similaridade de Vetor<span></span></h3>

Pequena explicação sobre o que é o Vetor

<h4> Vetores Aleatórios<span></span></h4>

Os vetores aleatórios (ou <i>`dummy vectors`</i>) são sequências de números gerados de forma aleatória. 
Esses números aqui, serão usados para simular os `embeddings` reais, e a forma como palavras e termos com significados semelhantes ficam próximas em um espaço vetorial.

```python
import numpy as np

for content in data:
    content['title_emb'] = np.random.rand(1536).astype(np.float32)  
```

O código cria um vetor aleatório para cada item de "data" e para cada conteúdo na lista de dados, ele gera um vetor com 1536 números aleatórios (entre 0 e 1) e os armazena como um novo campo chamado `'title_emb'`. O valor 1536 é uma dimensão padrão em modelos de geração de embeddings.

```python
import numpy as np

def index_documents_with_vectors(r, doc_prefix, documents, *vector_fields):
    for i, doc in enumerate(documents):
        key = f"{doc_prefix}:{i}"
        
        for field, value in doc.items():
            if isinstance(value, np.ndarray):
                doc[field] = value.tobytes()
            elif value is None:
                doc[field] = ""
            elif isinstance(value, str):
                doc[field] = value.encode('utf-8')
            elif isinstance(value, bytes):
                pass  # já está OK
            else:
                doc[field] = str(value).encode('utf-8') 

        r.hset(key, mapping=doc)
    
    print("Documents indexed.")
```

Para lidar com vetores, é necessário fazer alterações na forma com que o índice é criado. Essas mudanças incluem a transformação dos vetores, representados como arrays do NumPy, em bytes antes de serem armazenados no Redis, utilizando a função `tobytes().`

```python
import numpy as np
import redis
from redis.commands.search.field import TextField, NumericField, VectorField  

#nome do índice
index_name = 'idx:dummyTest'
# prefixo chaves para os documentos
doc_prefix = 'dummydoc:'  

vector_field = 'title_emb'  # nome do campo que contém os vetores
VECTOR_DIM = len(data[0][vector_field])  # dimensão dos vetores
VECTOR_NUMBER = len(data)  # número de documentos
DISTANCE_METRIC = "COSINE"  # métrica de distância (COSINE, IP, L2)

# defenição dos campos do índice no Redis
url = TextField(name="url")
title = TextField(name="title")
title_emb = VectorField(vector_field,
    "FLAT", {
        "TYPE": "FLOAT32",
        "DIM": VECTOR_DIM,
        "DISTANCE_METRIC": DISTANCE_METRIC,
        "INITIAL_CAP": VECTOR_NUMBER,
    }
)

fields = [url, title, title_emb]


# criação do indice no Redis
create_index(r, index_name, doc_prefix, fields)

# indexação dos documentos no Redis
index_documents_with_vectors(r, doc_prefix, data, vector_field)
```

Aqui é feita a criação do Índice apropriado para o uso de vetores, contendo campos de configuração e informação dos vetores e o campo `VectorField` que define os campos que existirão no índice.

<b> O KNN </b>

O KNN (K-Nearest Neighbors) é um algoritmo de aprendizado de máquina.
Aqui, ele será utilizado para encontrar documentos similares a um dado vetor, comparando distâncias dos vetores armazenados previamente no Redis.

```python
def search_redis(index_name, k, vector_field, return_fields, hybrid_fields = "*"):
   
    #get embeddings of a hyphotetical query
    query_emb = np.random.rand(1536)
    
    # Prepare the Query
    query = (
        Query(f"{hybrid_fields}=>[KNN {k} @{vector_field} $vector as score]")
         .return_fields(*return_fields)
         .sort_by("score")
         .paging(0, k)
         .dialect(2)
    )

    query_params = {"vector": np.array(query_emb, dtype=np.float32).tobytes()}

    # perform vector search
    results = r.ft(index_name).search(query, query_params)
    return results.docs
```

```python
k = 5
vector_field = "title_emb"
return_fields = ["id", "url", "title", "score"]
index_name = 'idx:dummyTest' #the name of the index, which you will use when doing queries

results = search_redis(index_name, k, vector_field, return_fields)
for doc in results:
    score = 1 - float(doc['score'])
    print(f"{doc['title']}; score:{score:.3f}")
    print(f"{doc['url']}")
    print("\n")
```

Esta etapa define a função `search_redis` e a executa no índice criado anteriormente, é gerado um vetor de consulta aleatório e a seguir devolve os 5 
documentos mais semelhantes de acordo com esse vetor, ordenados pela pontuação (ou <i>`score`</i>) de similaridade.

Com este processo, é possível confirmar o funcionamento dos vetores e seu cálculo, embora os resultados sejam aleatórios. Mesmo assim, é possível entender como o sistema lida com os vetores e como o processo de busca vetorial ocorre.

<h4> Vetores de Similaridade com BERT<span></span></h4>

```python
BERT ou SBERT (Bidirectional Encoder Representations from Transformers) é um modelo de linguagem usado para gerar representações de texto em 
forma de vetores de alta dimensão.
Aqui, ele será utilizado para criar vetores reais, que representem os títulos das notícias e realizar buscas por similaridade entre eles.
```

```python

!pip install redis numpy torch transformers --quiet
```

```python
#verificar daqui o que precisa de novo ou não:
import torch
from transformers import AutoTokenizer, AutoModel
```

```python
from transformers import AutoTokenizer, AutoModel

model_name = "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Adicionando from_tf=False
model = AutoModel.from_pretrained(model_name, from_tf=False)
```

Esse código carrega o modelo "paraphrase-multilingual-MiniLM-L12-v2" da biblioteca `transformers`.
Foi escolhido para a realização desta tarefa por ser adaptado para o português.

```python

def get_embeddings_bert(text, tokenizer, model):
    tokens = tokenizer(text, return_tensors="pt", padding=True, truncation=True)
    with torch.no_grad():
        output = model(**tokens)
    embeddings = output.last_hidden_state.mean(dim=1)
    return embeddings[0].numpy()
```

```python
Essa função gera embeddings (representações vetoriais) para um texto usando o modelo BERT, processando o texto com um tokenizador
e retornando a média dos vetores das últimas camadas do modelo.
```

```python
def create_vector_index(r, index_name, prefix, dim, vector_field):
    try:
        r.ft(index_name).info()
        print("Index already exists.")
    except:
        fields = [
            TextField("title"),
            TextField("summary"),
            TextField("url"),
            VectorField(vector_field,
                        "FLAT", {
                            "TYPE": "FLOAT32",
                            "DIM": dim,
                            "DISTANCE_METRIC": "COSINE"
                        })
        ]
        r.ft(index_name).create_index(fields=fields, definition=IndexDefinition(prefix=[prefix], index_type=IndexType.HASH))
        print("Vector index created.")
```

O novo índice é ajustado para armazenar vetores gerados por BERT, utilizando a métrica de similaridade `COSINE` e um vetor de dimensão especificada. 
Ele foi programado para armazenar os campos, `tittle`, `summary` e `url`, junto com o vetor de similaridade.

```python

def index_documents_with_bert_vectors(r, docs, prefix, vector_field):
    for i, doc in enumerate(data):
        emb = get_embeddings_bert(doc["title"] + " " + doc["summary"], tokenizer, model)
        r.hset(f"{prefix}{i}", mapping={
            "title": doc["title"],
            "summary": doc.get("summary", ""),
            "url": doc.get("url", ""),
            vector_field: emb.astype(np.float32).tobytes()
        })
    print("Documents indexed with embeddings.")
```

```python

#alterar para search_vector_redis, para não entrar em conflito com o anterior
def search_redis_bert(query_emb, index_name, k, vector_field, return_fields):
    query_vector = np.array(query_emb, dtype=np.float32).tobytes()
    base_query = f'*=>[KNN {k} @{vector_field} $vector AS score]'
    query = Query(base_query).return_fields(*return_fields).sort_by("score").dialect(2)
    params = {"vector": query_vector}
    results = r.ft(index_name).search(query, params)
    return results.docs
```

A função `index_documents_with_bert_vectors` indexa documentos no Redis, gerando embeddings com BERT para cada título e resumo,
e os armazena junto com outros campos como título, resumo e URL. Já a função `search_redis_bert` realiza uma busca no índice, 
comparando a consulta com os embeddings dos documentos utilizando KNN para encontrar os mais similares e retornando os resultados ordenados pelo score.
